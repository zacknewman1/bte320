```python
# Define the principal amount
principal = 3000

# Define the annual interest rate (as a decimal)
rate = 0.03

# Define the number of times interest is compounded per year (annually in this case)
n = 1

# Define the number of years the money is invested or borrowed for
t = 1 # Assuming 1 year initially, you can change this value

# Calculate the future value (A) using the compound interest formula: A = P(1 + r/n)^(nt)
amount = principal * (1 + rate / n)**(n * t)

# Calculate the compound interest
compound_interest = amount - principal

print(f"Principal: ${principal:.2f}")
print(f"Annual Interest Rate: {rate*100:.0f}%")
print(f"Time (years): {t}")
print(f"Future Value (Amount): ${amount:.2f}")
print(f"Compound Interest Earned: ${compound_interest:.2f}")
```
