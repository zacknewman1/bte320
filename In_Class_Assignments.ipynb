```python
sales = float(input("enter the total sales:"))
print(sales)
COGS = float(input("enter the total COGS:"))
print(COGS)
GP = sales - COGS
print(GP)
taxes = GP * 0.3
print(taxes)
net = GP - taxes
print(net)
```

```python
sales = float(input("enter the total sales ($):"))
cogs = .4 * sales
gross_profit = sales - cogs
print(f"Sales: {sales}$")
print(f"COGS: {cogs}$")
print(f"Gross Profit: {gross_profit}$")
```

```python
number1 = float(input("Enter the first number: "))
operator = input("Enter an operator (+, -, *, /): ")
number2 = float(input("Enter the second number: "))
if operator == "+":
  result = number1 + number2
elif operator == "-":
  result = number1 - number2
elif operator == "*":
  result = number1 * number2
elif operator == "/":
  result = number1 / number2
print(result)

Enter the first number: 14.3
Enter an operator (+, -, *, /): -
Enter the second number: 2
12.3
```
