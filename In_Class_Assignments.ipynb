```python
base_charge = 5
additional_hourly_rate = 2.50
min_fee = 10
max_fee = 20
hour = 1
print("Hour | Fee")
while hour <= 8:
    if hour == 1:
        fee = min_fee
    else:
        fee = base_charge + (hour - 1) * additional_hourly_rate
    fee = max(min_fee, fee)
    fee = min(max_fee, fee)
    hour += 1
    print(f"{hour - 1} | {fee}")
```

```python
import random
import gradio as gr
def generate_secret(start, end):
  return random.randint(int(start), int(end))
def start_game(start, end, attempts):
  start = int(start)
  end = int(end)
  attempts = int(attempts)

  secret = generate_secret(start, end)
  message = f"Game started! Guess a number between {start} and {end}."
  history = ""

  return secret, attempts, message, history

def evaluate_guess(guess, secret):
  guess = int(guess)
  secret = int(secret)

  if guess < secret:
    return "Too low!"
  elif guess > secret:
    return "Too high!"
  else:
    return "Correct!"

def update_history(history, guess):
  guess = str(int(guess))

  if history == "":
    return guess
  else:
    return history + " , " + guess
def process_guess(guess, secret, attempts_left, history):
  if attempts_left <= 0:
    return secret, attempts_left, "Game Over! Start a new game", history

  if guess is None:
    return secret, attempts_left, "Please enter a guess", history

  result = evaluate_guess(guess, secret)
  history = update_history(history, guess)

  if result == "Correct!":
    message = f"Correct! The number was {secret}"
    return secret, attempts_left, message, history


  attempts_left -= 1

  message = f"Incorrect! {result} Attempts left: {attempts_left}"

  if attempts_left == 0:
    message = f"Game Over! The number was {secret}"
    return secret, attempts_left, message, history

  return secret, attempts_left, message, history




# ----- GUI (DO NOT MODIFY) -----
with gr.Blocks() as demo:
  gr.Markdown("# Number Guessing Game")
  secret_state = gr.State(0)
  attempts_state = gr.State(0)
  history_state = gr.State("")
  with gr.Row(): # Components inside this block must be indented
    start_input = gr.Number(label="Range Start", value=1, precision=0)
    end_input = gr.Number(label="Range End", value=20, precision=0)
    attempts_input = gr.Number(label="Attempts", value=5, precision=0)
    start_button = gr.Button("Start Game")
    status_box = gr.Textbox(label="Game Status")
    history_box = gr.Textbox(label="Guess History")
  with gr.Row(): # Components inside this block must be indented
    guess_input = gr.Number(label="Your Guess", precision=0)
    guess_button = gr.Button("Submit Guess")
  start_button.click(
  fn=start_game,
  inputs=[start_input, end_input, attempts_input],
  outputs=[secret_state, attempts_state, status_box, history_state]
  )
  guess_button.click(
  fn=process_guess,
  inputs=[guess_input, secret_state, attempts_state, history_state],
  outputs=[secret_state, attempts_state, status_box, history_box]
  )
  demo.launch()
```

```python
def getValues (n):
  assert isinstance(n, int), "'n' must be an integer"
  items = []

  for i in range(n):
    items.append (float(input("Enter a value:")))

  if len(items) < 2:
    raise ValueError("the length of list 'items' cannot be < 2")
  return items


def calcAverage(alist):
  assert isinstance(alist, list), "'alist' must be a list"
  try:
    return sum(alist) / len(alist)
  except Exception as e:
    print(f"Error: {e}")


def main():
  n = int(input("'Enter a number of items to be added to the list: "))
  alist = getValues (n)
  avg = calcAverage(alist)
  print (f"Average weight: {avg} lbs.")
main ()
```

```python
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    def display(self, formal=False):
        if formal:
            print(f"{self.last_name}, {self.first_name}")
        else:
            print(self.first_name)

    def __str__(self):
        return f"Instance of class {self.__class__.__name__}"

name = Person ("John" , "Doe")
print(name)
name.display()
name.display(formal=True)
```

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        return math.sqrt((self.x - other.x)**2 + (self.y - other.y)**2)

class Rectangle:
    def __init__(self):
        self.p1 = Point(0,0)
        self.p2 = Point(3,0)
        self.p3 = Point(0,3)
        self.p4 = Point(3,3)
        self.width = self.p1.distance(self.p2)
        self.height = self.p1.distance(self.p3)

    def area(self):
        return self.width * self.height

z = Rectangle()
z.area()
```

```python
class Employee:
    def __init__(self, name, wage, hours):
        self.name = name
        self.__hourly_wage = wage
        self.hours = hours

    def get_wage(self):
      return self.__hourly_wage

    def set_wage(self, newWage):
      if newWage <= 0:
        print("Invalid wage")
      else:
        self.__hourly_wage = newWage

    def calculate_pay(self):
        return self.__hourly_wage * self.hours

    def display_info(self):
        print(f"Employee: {self.name}")
        print(f"Hourly wages: ${self.__hourly_wage}")
        print(f"Hours worked: {self.hours}")
        print(f"Total Pay: ${self.calculate_pay()}")

    def __str__(self):
        return f"Employee: {self.name}\nHourly wages: {self.__hourly_wage}"

class Manager(Employee):
    def __init__(self, name, hourly_wage, hours_worked, bonus):
        super().__init__(name, hourly_wage, hours_worked)
        self.bonus = bonus

    def calculate_pay(self):
        base_pay = super().calculate_pay()
        return base_pay + self.bonus

    def display_info(self):
        super().display_info()
        print(f"Bonus: ${self.bonus}")

print("Employee Information:")
employee = Employee("Kique", 20, 40)
employee.display_info()

print("Manager Information:")
manager = Manager("Luke", 30, 40, 500)
manager.display_info()
```
